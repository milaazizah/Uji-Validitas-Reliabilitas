```python
import re
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files
```

```python
up = files.upload()
nama_file = list(up.keys())[0]
print("File terbaca:", nama_file)
```

```python
JUMLAH_KOLOM_DEMOGRAFI = 6

df = pd.read_excel(nama_file)
butir = df.iloc[:, JUMLAH_KOLOM_DEMOGRAFI:]

print("Jumlah responden :", len(df))
print("Jumlah butir :", butir.shape[1])
print("Data kosong :", int(butir.isna().sum().sum()))
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5,
}

num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")

# Harus 0. Jika lebih dari 0, ada jawaban yang tidak terbaca.
print("Sel yang gagal dikonversi:", int(num.isna().sum().sum()))
num.head()
```

```python
import re

dimensi = {}
for kolom in butir.columns:
    m = re.match(r"^(.*?) \[", kolom)
    if m is None:
        print("Dilewati:", kolom)
        continue
    dimensi.setdefault(m.group(1).strip(), []).append(kolom)

for nama, daftar in dimensi.items():
    print(f"{nama}: {len(daftar)} butir")

butir = butir[[k for daftar in dimensi.values() for k in daftar]]
num = num[butir.columns]
```

```python
n = len(num)
t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)
print(f"n = {n} | r tabel (α=0,05; df={n-2}) = {r_tabel:.3f}")

hasil_validitas = []
for nama, daftar in dimensi.items():
    x = num[daftar]
    total = x.sum(axis=1)
    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)
        r_koreksi, _ = stats.pearsonr(x[kolom], total - x[kolom])
        hasil_validitas.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value": round(p, 4),
            "Keterangan": "Valid" if r > r_tabel else "Tidak valid",
        })

tabel_validitas = pd.DataFrame(hasil_validitas)
tabel_validitas
```

```python
def cronbach_alpha(x):
    k = x.shape[1]
    return k / (k - 1) * (1 - x.var(ddof=1).sum() / x.sum(axis=1).var(ddof=1))

def kategori(a):
    return ("Sangat baik" if a >= 0.90 else "Baik" if a >= 0.80
            else "Reliabel" if a >= 0.70 else "Tidak reliabel")

hasil_reliabilitas = []
alpha_dihapus = []

for nama, daftar in dimensi.items():
    x = num[daftar]
    a = cronbach_alpha(x)
    hasil_reliabilitas.append({"Dimensi": nama, "Jumlah butir": len(daftar),
                               "Cronbach's alpha": round(a, 3), "Kategori": kategori(a)})
    for i, kolom in enumerate(daftar, 1):
        if len(daftar) > 2:
            alpha_dihapus.append({"Dimensi": nama, "Butir": f"B{i}",
                                  "Alpha jika dihapus": round(cronbach_alpha(x.drop(columns=kolom)), 3)})

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)

display(tabel_reliabilitas)
display(tabel_alpha_dihapus)
```

```python
gabung = tabel_validitas.merge(tabel_alpha_dihapus, on=["Dimensi", "Butir"], how="left")

with pd.ExcelWriter("hasil_uji_validitas_reliabilitas.xlsx") as w:
    gabung.to_excel(w, sheet_name="Validitas", index=False)
    tabel_reliabilitas.to_excel(w, sheet_name="Reliabilitas", index=False)

files.download("hasil_uji_validitas_reliabilitas.xlsx")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Butuh variabel dari langkah sebelumnya: dimensi (dict) dan num (data angka 1-5)
n = len(num)
se_skew = np.sqrt(6*n*(n-1)/((n-2)*(n+1)*(n+3)))   # galat baku skewness

label = ["STS", "TS", "N", "S", "SS"]
warna = ["#B03A2E", "#E59866", "#BFC9CA", "#5DADE2", "#1F4E79"]

# ---------- 1. Tabel statistik deskriptif per dimensi ----------
hasil = []
for nama, daftar in dimensi.items():
    skor = num[daftar].mean(axis=1)            # skor dimensi = rata-rata butir (skala 1-5)
    hasil.append({
        "Dimensi": nama.strip(" ."),
        "Rata-rata": round(skor.mean(), 2),
        "Std. Deviasi": round(skor.std(ddof=1), 2),
        "Skewness": round(skor.skew(), 2),
        "z Skewness": round(skor.skew()/se_skew, 2),   # |z| > 1,96 -> kemiringan signifikan
        "Kurtosis": round(skor.kurt(), 2),             # excess kurtosis (normal = 0)
    })
tabel_deskriptif = pd.DataFrame(hasil)
display(tabel_deskriptif)

# ---------- 2. Grafik per dimensi: komposisi jawaban + sebaran skor dimensi ----------
for nama, daftar in dimensi.items():
    skor = num[daftar].mean(axis=1)
    persen = pd.DataFrame({
        f"B{i}": num[k].value_counts(normalize=True).reindex(range(1, 6), fill_value=0) * 100
        for i, k in enumerate(daftar, 1)}).T
    persen.columns = label

    fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 0.6*len(daftar) + 2.6),
                                 gridspec_kw={"width_ratios": [1.6, 1]})

    # kiri: komposisi jawaban per butir (%)
    persen.plot(kind="barh", stacked=True, color=warna, width=0.8, ax=a1, legend=False)
    for bar_group, col in zip(a1.containers, label):
        a1.bar_label(bar_group, labels=[f"{v:.0f}" if v >= 7 else "" for v in persen[col]],
                     label_type="center", fontsize=8,
                     color="white" if col in ("STS", "SS") else "#222")
    a1.invert_yaxis(); a1.set_xlim(0, 100)
    a1.set_xlabel("Persentase responden (%)"); a1.set_title("Komposisi jawaban per butir")
    a1.legend(ncol=5, loc="upper center", bbox_to_anchor=(0.5, -0.18), frameon=False)

    # kanan: histogram skor dimensi + SD, skewness, kurtosis
    a2.hist(skor, bins=np.arange(1, 5.51, 0.5), color="#5DADE2", edgecolor="white")
    a2.axvline(skor.mean(), color="#1F4E79", ls="--", lw=1.5)
    a2.set_xlim(1, 5); a2.set_xlabel("Skor dimensi (rata-rata butir)"); a2.set_ylabel("Jumlah responden")
    a2.set_title("Sebaran skor dimensi")
    teks = (f"Rata-rata = {skor.mean():.2f} (garis putus-putus)\nSD = {skor.std(ddof=1):.2f}\n"
            f"Skewness = {skor.skew():.2f}\nKurtosis = {skor.kurt():.2f}")
    a2.text(0.03, 0.97, teks, transform=a2.transAxes, va="top", fontsize=10,
            bbox=dict(boxstyle="round", fc="white", ec="#BBBBBB"))

    fig.suptitle(nama.strip(" ."), fontweight="bold")
    fig.tight_layout()
    plt.show()
```
